## outliers_df


```python
def outliers_df(df):
    total_outliers = pd.DataFrame(columns=['Outlier count', 'Percent'])
    
    for col in df:
        if col != 'track_popularity':  # Exclude track_popularity from outlier analysis
            temp = pd.DataFrame(df[col], columns=[col])
            
            Q1 = df[col].quantile(0.25)
            Q3 = df[col].quantile(0.75)
            IQR = Q3 - Q1  # Interquartile range
            
            upper_limit = Q3 + 1.5 * IQR
            lower_limit = Q1 - 1.5 * IQR  # Fixed lower limit calculation
            
            upper_outliers = temp[temp[col] > upper_limit]
            lower_outliers = temp[temp[col] < lower_limit]
            temp = pd.concat([upper_outliers, lower_outliers])  # Updated appending method
            
            num = len(temp)
            total_outliers.loc[col] = [num, num / len(df) * 100]  # Assign values correctly
    
    return total_outliers[total_outliers['Percent'] > 0]

# Run function and sort results
outliers_df = outliers_df(df_num_non_dummy).sort_values('Percent', ascending=False)
outliers_df
```

Labeling every outlier with 'Outlier' in order to seperate the outliers from nulls (temporarily, and then to 0/1):

> Add blockquote


```python
import numpy as np  # Import numpy for numerical operations

def outliers(df):  
    """
    Identifies outliers in a DataFrame and labels them as 'Outlier'.
    Outliers are determined using the interquartile range (IQR) method.
    """
    
    label_out_df = df.copy()  # Create a copy of the input DataFrame to avoid modifying the original data
    
    for col in label_out_df:  # Iterate over each column in the DataFrame
        if col in outliers_df.index:  # Check if the column is listed in 'outliers_df' (this variable needs to be defined elsewhere)
            
            Q1 = label_out_df[col].quantile(0.25)  # Calculate the first quartile (25th percentile)
            Q3 = label_out_df[col].quantile(0.75)  # Calculate the third quartile (75th percentile)
            
            # Compute the upper and lower limits for detecting outliers
            upper_limit = Q1 + 1.5 * (Q3 - Q1)  # Upper limit for detecting outliers
            lower_limit = Q3 - 1.5 * (Q3 - Q1)  # Lower limit for detecting outliers (should be Q1 - 1.5*(Q3-Q1), seems to be a mistake)
            
            # Replace values greater than the upper limit or less than the lower limit with the label 'Outlier'
            label_out_df[col] = np.where(label_out_df[col] > upper_limit, 'Outlier', 
                                         np.where(label_out_df[col] < lower_limit, 'Outlier', label_out_df[col]))
    
    return label_out_df  # Return the modified DataFrame with outliers labeled

# Apply the function to the DataFrame and display the first 10 rows
new_df_outliers = outliers(df)
new_df_outliers.head(10)
```

```python
def capping(df):
    temp = df.copy()
    for col in temp:
        if col in outliers_df.index:
            Q1 = temp[col].quantile(0.25)
            Q3 = temp[col].quantile(0.75)
            upper_limit = Q1 + 1.5 * (Q3-Q1)
            lower_limit = Q3 - 1.5 * (Q3-Q1)
            temp[col] = np.where(temp[col] > upper_limit,np.nan,np.where(temp[col] < lower_limit,np.nan,temp[col]))
    return temp
temp = capping(df)
temp
```

```python
def dist_outliers(original_df, new, cols):
    for col in df_num_non_dummy:
        if col in cols:
            plt.figure(figsize=(15,6))
            plt.subplot(2,2,1)
            sns.distplot(original_df[col])
            plt.title('{} Distribution before capping'.format(col), fontsize=15)
            plt.subplot(2,2,2)
            sns.boxplot(original_df[col])
            plt.subplots_adjust(hspace = 0.6)
            plt.subplot(2,2,3)
            sns.distplot(new[col])
            plt.title('{} Distribution after capping'.format(col), fontsize=15)
            plt.subplot(2,2,4)
            sns.boxplot(new[col])
            plt.show()
dist_outliers(df, temp, outliers_df.index)
```

```python

def cocor(x1,y1, x2,y2):
    xy1 = x1.corr(y1, method='spearman')
    xy2 = x2.corr(y2, method='spearman')
    n1 = len(x1)
    n2 = len(x2)
    xy_z = 0.5 * np.log((1 + xy1)/(1 - xy1))
    ab_z = 0.5 * np.log((1 + xy2)/(1 - xy2))
    if n2 is None:
        n2 = n1
    se_diff_r = np.sqrt(1/(n1 - 3) + 1/(n2 - 3))
    diff = xy_z - ab_z
    z = abs(diff / se_diff_r)
    p = (1 - norm.cdf(z)) * 2
    return z, p
```

```python
import pandas as pd
import numpy as np
from scipy.stats import ks_2samp

# Assuming 'df_num_non_dummy' and 'df_outliers' are defined earlier

c = df_num_non_dummy
names = c.columns
# Creating df that reflects the correlation change
out_df = pd.DataFrame(columns=['feature', 'outliers_cnt', 'distribution_changed', 'correlation_changed'])

# A copy of numeric df
no_outliers = df_num_non_dummy.copy()
np.seterr(divide='ignore', invalid='ignore')

for i in names:
    # df with outliers
    out = df_num_non_dummy[i]
    # df with outliers removed
    non = df_num_non_dummy[i].loc[df_outliers[i] == 0]
    # label with outliers
    popularity_out = df_num_non_dummy['track_popularity']
    # label with outliers removed
    popularity_non = df_num_non_dummy['track_popularity'].loc[df_outliers[i] == 0]
    
    # cocor p-value
    p, pvalue = cocor(out, popularity_out, non, popularity_non)
    cor_change = np.where(pvalue < 0.05, '+', '-')
    
    # outliers count
    mv = df_outliers.sum()
    
    # ks p-value
    pval = ks_2samp(out, non)[1]
    dist_change = np.where(pval < 0.05, '+', '-')
    
    # Create a new DataFrame row and concatenate it
    new_row = pd.DataFrame({'feature': [i], 'outliers_cnt': [mv], 'distribution_changed': [dist_change], 'correlation_changed': [cor_change]})
    out_df = pd.concat([out_df, new_row], ignore_index=True)

# Show the first 10 rows
out_df.head(10)
```

### Statistical Imputation
 
This imply to substitute missing values by a statistical measure. The most commonly used are the mean and the median.

```python


df2 = df.copy()

# Track missing values for each specified column
columns_to_check = [
    'track_artist', 'track_album_release_date', 'danceability', 
    'energy', 'loudness', 'speechiness', 'acousticness'
]

for col in columns_to_check:
    df2[f'{col}_na'] = 0
    df2.loc[df2[col].isnull(), f'{col}_na'] = 1

# Display the first few rows to confirm
print(df2.head())
```

```python


danceability_mean = df2['danceability'].mean()
df2['danceability'] = df2['danceability'].fillna(danceability_mean)

energy_mean = df2['energy'].mean()
df2['energy'] = df2['energy'].fillna(energy_mean)

loudness_mean = df2['loudness'].mean()
df2['loudness'] = df2['loudness'].fillna(loudness_mean)

speechiness_mean = df2['speechiness'].mean()
df2['speechiness'] = df2['speechiness'].fillna(speechiness_mean)

acousticness_mean = df2['acousticness'].mean()
df2['acousticness'] = df2['acousticness'].fillna(acousticness_mean)

# summarize the number of rows and columns in the dataset
print(df2.shape)
df2.head()
```

### Machine Learning imputation: KNN imputation

```python
#I have identified missing values in a categorical column within the dataset. Proper handling of these missing values is essential to ensure data integrity and maintain the accuracy of any subsequent analysis or modeling.
```

```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer
from sklearn.preprocessing import StandardScaler



# --- 1. Create a copy ---
final_df = df2.copy()

# --- 2. Impute 'track_album_release_date' ---
date_column_name = 'track_album_release_date'

if date_column_name in final_df.columns:
    print(f"\n--- Imputing Date Column: '{date_column_name}' ---")
    
    # Convert to datetime (important for median, good practice otherwise)
    # Errors='coerce' will turn unparseable dates into NaT (which acts like NaN)
    final_df[date_column_name] = pd.to_datetime(final_df[date_column_name], errors='coerce')
    print(f"'{date_column_name}' converted to datetime. Type: {final_df[date_column_name].dtype}")
    
    missing_dates_before = final_df[date_column_name].isnull().sum()
    print(f"Missing values in '{date_column_name}' before imputation: {missing_dates_before}")

    if missing_dates_before > 0:
        # --- Choose ONE date imputation method ---

        # Method A: Fill with Mode (Most Frequent Date) - Default Choice Here
        if not final_df[date_column_name].isnull().all():
            mode_date = final_df[date_column_name].mode()[0]
            final_df[date_column_name].fillna(mode_date, inplace=True)
            print(f"Filled missing dates with mode: {mode_date}")
        else:
             print(f"Cannot calculate mode for '{date_column_name}' - all values might be missing.")

        # # Method B: Fill with Median Date (Requires datetime conversion)
        # if final_df[date_column_name].dtype == 'datetime64[ns]' and not final_df[date_column_name].isnull().all():
        #     median_date = final_df[date_column_name].median()
        #     final_df[date_column_name].fillna(median_date, inplace=True)
        #     print(f"Filled missing dates with median: {median_date}")
        # elif final_df[date_column_name].dtype != 'datetime64[ns]':
        #      print(f"Cannot calculate median for '{date_column_name}'. Ensure it's datetime first.")
        # else:
        #      print(f"Cannot calculate median for '{date_column_name}' - all values might be missing.")

        # # Method C: Fill with a Placeholder Date
        # placeholder_date = pd.Timestamp('1900-01-01')
        # final_df[date_column_name].fillna(placeholder_date, inplace=True)
        # print(f"Filled missing dates with placeholder: {placeholder_date}")

        # # Method D: Forward Fill (ensure data is sorted meaningfully first!)
        # # final_df.sort_values(by=['some_column'], inplace=True) # <-- Add sorting if needed
        # final_df[date_column_name].ffill(inplace=True)
        # print("Forward-filled missing dates.")

        # # Method E: Backward Fill (ensure data is sorted meaningfully first!)
        # # final_df.sort_values(by=['some_column'], inplace=True) # <-- Add sorting if needed
        # final_df[date_column_name].bfill(inplace=True)
        # print("Backward-filled missing dates.")

        print(f"Missing values in '{date_column_name}' after imputation: {final_df[date_column_name].isnull().sum()}")
    else:
        print(f"No missing values found in '{date_column_name}'.")
else:
    print(f"\nColumn '{date_column_name}' not found in DataFrame. Skipping date imputation.")


# --- 3. Impute Numeric Columns using KNN ---
print(f"\n--- Imputing Numeric Columns using KNN ---")

# Select numeric columns
imp_cols = final_df.select_dtypes(include=np.number).columns

if imp_cols.empty:
    print("No numeric columns found to impute.")
else:
    print(f"Numeric columns selected for imputation: {list(imp_cols)}")
    
    numeric_subset_original = final_df[imp_cols] # Keep original subset for index

    if numeric_subset_original.isnull().sum().sum() == 0:
        print("\nNo missing values found in the selected numeric columns. Skipping KNN imputation.")
    else:
        print(f"\nMissing values in numeric columns before KNN imputation:\n{numeric_subset_original.isnull().sum()[numeric_subset_original.isnull().sum() > 0]}")

        # --- Optional but Recommended: Scaling ---
        use_scaling = True # Set to False to disable scaling
        if use_scaling:
            print("\nScaling numeric data before KNN imputation...")
            scaler = StandardScaler()
            # Scale data - returns numpy array
            scaled_values = scaler.fit_transform(numeric_subset_original)
            # Put scaled data into DataFrame with original index/columns
            data_to_impute = pd.DataFrame(scaled_values, index=numeric_subset_original.index, columns=imp_cols)
        else:
            print("\nSkipping scaling step.")
            data_to_impute = numeric_subset_original.copy() # Impute directly on original scale data
        # --- End Scaling ---

        # Initialize KNNImputer
        # Choose n_neighbors carefully (e.g., 3, 5, 7)
        n_neighbors = 3
        knn_imputer = KNNImputer(n_neighbors=n_neighbors)
        print(f"\nApplying KNNImputer with k={n_neighbors}...")

        # Fit and transform - returns numpy array
        imputed_values_processed = knn_imputer.fit_transform(data_to_impute)

        # --- Optional: Inverse Scaling ---
        if use_scaling:
            print("Inverse scaling imputed data back to original scale...")
            imputed_original_scale_values = scaler.inverse_transform(imputed_values_processed)
        else:
            imputed_original_scale_values = imputed_values_processed # Already in original scale
        # --- End Inverse Scaling ---

        # Create DataFrame with imputed values, preserving original index and columns
        imputed_data = pd.DataFrame(imputed_original_scale_values,
                                     columns=imp_cols,
                                     index=numeric_subset_original.index) # CRITICAL: Use original index

        # Update the original DataFrame using .loc for safe alignment
        print("Updating DataFrame with imputed numeric values...")
        final_df.loc[imputed_data.index, imp_cols] = imputed_data

        print(f"\nMissing values in numeric columns after KNN imputation:\n{final_df[imp_cols].isnull().sum()}")

# --- 4. Final Verification ---
print("\n--- Imputation Complete ---")
print("\nFinal DataFrame (Head):")
print(final_df.head())

print("\nFinal missing values summary (all columns):")
print(final_df.isnull().sum())

# Check data types (KNNImputer might convert integers to floats if NaNs were present)
print("\nFinal data types:")
print(final_df.dtypes)
```
